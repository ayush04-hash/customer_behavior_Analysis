## 2. Clean missing values

```python
# check missing values
df.isnull().sum()
```

```python
# fill missing Review Rating with the median rating for that product Category
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
# confirm no missing values remain
df.isnull().sum()
```

## 3. Standardize column names to snake_case

```python
# step 1: convert to lower case
df.columns = df.columns.str.lower()

# step 2: convert spaces to underscores
df.columns = df.columns.str.replace(' ', '_')

# step 3: rename column
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})

df.columns
```

## 4. Feature engineering

```python
# create an age_group column to simplify pattern analysis

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
# qcut divides data into equal-sized buckets based on rank/distribution

df[['age', 'age_group']].head(10)
```

```python
# create a purchase_frequency_days column
# converts frequency_of_purchases text into a numeric value (days) to simplify analysis

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-weekly': 14,
    'Annually': 365,   # FIXED: was misspelled 'Annualy', which mapped every Annually row to NaN
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python
# confirm the mapping now covers every category with no leftover NaNs
df['purchase_frequency_days'].isnull().sum()
```

## 5. Drop redundant column

```python
# check whether discount_applied and promo_code_used always match
(df['discount_applied'] == df['promo_code_used']).all()
# both show the same values, so we don't need 2 columns - one is sufficient
# we will delete the promo_code_used column
```

```python
df = df.drop('promo_code_used', axis=1)
df.columns
```
